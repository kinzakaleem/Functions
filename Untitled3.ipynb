1. What is a difference between a function and a method in python
- A function is a block of reusable code that is defined using the def keyword and can be called independently. Functions are typically defined at the module level (outside any class).
A method is a function that is associated with a class and is called on an instance(or sometimes in the class itself). It typically operates on the data(attributes) of the object it is called on. It is defined within a class and usually requires atleast one parameter.

2. Explain the concept of function arguments and parameters in python

4. What is the purpose of the return statement in a python
- The return statement in python serves the purpose of ending a fucntion and optionally sending a value(lor values) back to the caller. It allows functions to produce a result that can be stored, used , or processed further.

5. What are iterates in python and how do they differ from iterables

An Iterable is any object in python that can return its elements one at a time.

6. Expalin the concept of generators in python and how they are defined
- Generators are special type of iterable in python that allow you to create iterators in a concise, memory efficient way. Instead of computing all the values at once and storing them in memory (as lists or tuples do), generators yield item one at a time as they are required

7. What are the advantages of using generators over regular function
- Generators offer several advantages over regular functions, particularly when dealing with large datasets or streams of data. Memory efficiency, lazy evaluation, infinite sequence, improved perfomence,  simpler syntax for complex iterators.

8. What is a lambda function in python and when is it typically used
- A lambda function is an anonymous, single expression function defined using the lambda keyword. Unlike regular functions created with def, lambda functions do not ahve a name (unless assigned to a variable) and are limited to a single expression, which is implicity returned.

9. Explain the purpose and usage of a map function in python.

It allows for concise and efficient transformation or processing of data without the need for explicit loops.

```python
# a python function that takes a list of number as input and return the of all even numbers in

def sum_of_even_numbers(numbers):
    """
    returns the sum of all even numbers in the input list.

    parameters:
    numbers (list): a list of numbers

    returns:
    int: the sum of all even numbers in the input list
    """
    return sum([num for num in numbers if num % 2 == 0])
numbers = [1, 2, 3, 4, 5, 6]
result = sum_of_even_numbers(numbers)
print(result)
```

```python
# create a python function  that accepts a string and returns the reverse of that string

def reverse_string(input_string):
  """
  returns the reverse of the input string

  parameters:
  input_string (str): the string to be reversed

  returns:
  str: the reversed string
  """
  return input_string[::-1]

input_string = "hello world"
result = reverse_string(input_string)
print(result)
```

```python
# a python fucntion that takes a list of integers and returns a new list containing the square of each number

def square_numbers(numbers):
  """
  returns a new list containing the square of each number in the input list

  parameters:
  numbers (list): a list of integers

  returns:
  list: a new list containing the square of each number in the input list
  """
  return [num**2 for num in numbers]

numbers = [1, 2, 3, 4, 5]
result = square_numbers(numbers)
print(result)
```

```python
# a python function that checks if a given number is prime or not 1 to 200

def is_prime(number):
  """
  checks if a given number is prime or not

  parameters:
  number (int): the number to be checked

  returns:
  bool: True if the number is prime, False otherwise
  """
  if number < 2:
    return False
  for i in range(2, int(number**0.5) + 1):
    if number % i == 0:
      return False
  return True

for number in range(1, 201):
  if is_prime(number):
    print(number)
```

```python
# an iterator class in python that generates the fibanoccci sequence up to a specified numbers of atoms

class FibonacciIterator:
  """
  an iterator class in python that generates the fibanoccci sequence up to a specified numbers of atoms
  """
  def __init__(self, num_terms):
    """
    initializes the fibonacci iterator with the number of terms to generate

    parameters:
    num_terms (int): the number of terms to generate in the fibonacci sequence
    """
    self.num_terms = num_terms
    self.current_term = 0
    self.next_term = 1
    self.index = 0
  def __iter__(self):
        return self
  def __next__(self):
    if self.index >= self.num_terms:
      raise StopIteration
    else:
      fib_name = self.current_term
      self.current_term, self.next_term = self.next_term, self.current_term + self.next_term
    self.index += 1
    return fib_name

num_terms = 10
fib_iter = FibonacciIterator(num_terms)
for num in fib_iter:
  print(num)
```

```python
# a generator function in python that yields the powers of 2 up to a given exponent

def powers_of_two(exponent):
  """
  a generator function in python that yields the powers of 2 up to a given exponent

  parameters:
  exponent (int): the maximum exponent for which to compute powers of 2

  yields:
  int: the next power of 2 from 2^0 up to 2^exponent
  """
  for i in range(exponent + 1):
    yield 2 ** i

exponent = 5
for power in powers_of_two(exponent):
  print(power)
```

```python
# generate function that reads a file line by line and yields each line as string

def read_file_line_by_line(file_path):
  """
  a generator function that reads a file line by line and yields each line as a string

  parameters:
  file_path (str): the path to the file to read

  yields:
  str: each line in the file
  """
  with open(file_path, 'r') as file:
    for line in file:
      yield line.strip() # using strip() to remove the new line character
# example usage:
file_path = 'example.txt' # replace with the path to your file
for line in read_file_line_by_line(file_path):
  print(line)
```

```python
# a python program that uses the map funtion to convert temp from celsious to fahrenhit

def celsius_to_fahrenhit(celsius):
  return (celsius * 9/5) + 32

celsius_temperatures = [0, 10, 20, 30, 40]
fahrenhit_temperatures = list(map(celsius_to_fahrenhit, celsius_temperatures))
print(fahrenhit_temperatures)
```
